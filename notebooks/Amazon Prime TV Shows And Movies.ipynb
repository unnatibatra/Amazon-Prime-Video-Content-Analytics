## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Handling the duplicates
df.drop_duplicates(inplace=True)
````

````python
df.duplicated().sum()
````

````python
# Handling Missing Values
df.isnull().sum()
````

````python
#description
df['description'] = df['description'].fillna('Description Not Available')
````

````python
df['description'].isnull().sum()
````

````python
# age certification
df['age_certification'].mode()
````

````python
df['age_certification'].fillna('R',inplace=True)
df['age_certification'].isnull().sum()
````

````python
print(df.columns.tolist())
````

````python
# seasons
pd.crosstab(df['type'], df['seasons'].isna())
````

````python
df['seasons'] = df['seasons'].fillna(0)
````

````python
df['seasons'].isnull().sum()
````

````python
# imdb_id : as it is unique for everyone, we can't replace with mean/mode/median
df['imdb_id'].fillna('unknown', inplace=True)
df.isnull().sum()
````

````python
# imdb_score
df['imdb_score'].describe()
````

````python
df['imdb_score'] = df['imdb_score'].fillna(df['imdb_score'].median())
df['imdb_score'].isna().sum()
````

````python
# imdb_votes
print("Mean:", df['imdb_votes'].mean())
print("Median:", df['imdb_votes'].median())
````

````python
sns.histplot(df['imdb_votes'].dropna(), kde=True)
plt.show()
````

````python
df['imdb_votes'] = df['imdb_votes'].fillna(0)
df['imdb_votes'].isna().sum()
````

````python
# tmdb_popularity: we can drop as it has only 15 missing values
df.dropna(subset=['tmdb_popularity'],inplace =True)
df.isnull().sum()
````

````python
# tmdb_score
df['tmdb_score'].describe()
````

````python
plt.figure(figsize=(10,5))
sns.histplot(df,x='tmdb_score',bins=20,kde=True)
plt.show()
````

````python
df['tmdb_score'] = df['tmdb_score'].fillna(df['tmdb_score'].mean())
df.isnull().sum()
````

````python
# character
df['character'].mode()
````

````python
df['character'].fillna('Himself', inplace=True)
df['character'].isnull().sum()
````

### What all manipulations have you done and insights you found?

Data inspection & duplicates: Examined the dataset structure, data types, duplicate records, and missing-value counts before beginning EDA. Duplicate records were identified and handled to avoid affecting the analysis.


Seasons — 116,026 missing values: Since this field represents the number of seasons for a TV show, I checked it against the content type and treated missing values for movies as 0 (not applicable).


imdb_score — 6,051 missing values: Compared the mean and median and analyzed the distribution. Since the median was 6.1 and was less affected by extreme values, missing scores were replaced with the median.


imdb_votes — 6,075 missing values: The variable was highly right-skewed (mean ≈ 23,275 vs median ≈ 1,257). Since missing votes do not necessarily represent zero votes, the missing values were retained rather than artificially imputed.


tmdb_popularity — 15 missing values: Since only 15 observations were missing out of the entire dataset, these records were removed, as the proportion was negligible.


tmdb_score — 10,154 missing values: The distribution was approximately symmetric, with mean ≈ 5.99 and median ≈ 6.10. Therefore, missing values were replaced using mean imputation.


character — 16,277 missing values: Since character is a categorical variable representing character names, mode imputation was considered for the missing categorical values.


EDA preparation: After cleaning, the data was prepared for univariate, bivariate, and multivariate analysis to investigate content type, genres, release trends, ratings, popularity, and other factors influencing Amazon Prime's content library.
